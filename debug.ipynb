```python
import os
import sys
```

```python
base_path = !pwd
```

```python
dummy_dir_name = "DUMMY_dir"
```

```python
dummy_dir_path = base_path[0] + '/' + dummy_dir_name
```

```python
if not os.path.exists(dummy_dir_path):
    try:
        os.makedirs(dummy_dir_path)
        os.chmod(dummy_dir_path, 0o777) #need to add write permissions here, equivalent to chmod 777 from CLI 
        print("Created DUMMY dir at: ", dummy_dir_path)
    except:
        print('   ERROR: output dir {} doesn\'t exist: please create it first!'.format(dummy_dir_path))
        print("Unexpected error:", sys.exc_info()[0])
        print(sys.exit(2))
```
